```python
stock_symbol = "GE"
aapl2023 = yf.Ticker(stock_symbol).history(period="1y")
a = aapl2023[["High", "Low"]].reset_index().melt(id_vars="Date")
plt.figure(figsize=(10, 6))
sb.lineplot(data=a, x="Date", y="value", hue="variable")
divs = aapl2023[["Stock Splits", "Dividends"]].reset_index().melt(id_vars="Date")
divs = divs[divs["value"] != 0.0]
divs["y"] = a.min()["value"]
sb.scatterplot(data=divs, x="Date", y="y", hue="variable")
```

```python
def hilo_and_divs(df, name=None):
    hl = df[["High", "Low"]].reset_index().melt(id_vars="Date")
    divs = df[["Stock Splits", "Dividends"]].reset_index().melt(id_vars="Date")
    divs = divs[divs["value"] != 0.0]
    divs["y"] = hl.min()["value"]
    hl["Name"] = name
    divs["Name"] = name
    return hl, divs
```

```python
ge = hilo_and_divs(yf.Ticker("GE").history(period="1y"), name=yf.Ticker("GE").info.get("shortName"))
aapl = hilo_and_divs(yf.Ticker("AAPL").history(period="1y"), name=yf.Ticker("AAPL").info.get("shortName"))
hl = pd.concat([ge[0], aapl[0]])
divs = pd.concat([ge[1], aapl[1]])
divs.y = divs.min().y
plt.figure()
fig = sb.relplot(data=hl, kind="line", col="Name", x="Date", y="value", hue="variable")
for ax in fig.axes[0]:
    sb.scatterplot(data=divs[divs["Name"] == ax.title.get_text()[7:]], x="Date", y="y", hue="variable", ax=ax)
```

```python
dji = "MMM AXP AMGN AMZN AAPL BA CAT CVX CSCO KO DIS GS HD HON IBM JNJ JPM MCD MRK MSFT NKE NVDA PG CRM SHW TRV UNH VZ V WMT".split(" ")
dji_t =  yf.Tickers(" ".join(dji)).tickers
holders = pd.concat(map(lambda x: dji_t[x].institutional_holders, dji))
holders.value_counts("Holder")
```

```python
holders.groupby("Holder").sum("Value")["Value"].sort_values(ascending=False).head(10).plot(kind="bar")
```

```python
sector = pd.DataFrame(data=list(map(lambda x: (x, dji_t[x].info.get("sector")), dji)))
sector.value_counts(1).plot(kind="pie")
```

```python
def get_esg(code):
    return dji_t[code].sustainability.loc["totalEsg"]["esgScores"]
sector["sus"] = sector[0].apply(get_esg)
sb.barplot(data=sector, y=1, x="sus", errorbar=None, orient="h")
```

```python
mmm = dji_t["MMM"].history("10y")
mmm["yy-mm"] = pd.to_datetime(mmm.index.strftime('%Y-%m'))
sb.lineplot(data=mmm, x="yy-mm", y="High", errorbar=None)
```

```python
dji10y = []
for code in dji:
    df = dji_t[code].history("10y")
    df["yy-mm"] = pd.to_datetime(df.index.strftime('%Y-%m'))
    df["Code"] = code
    df["Sector"] = dji_t[code].info.get("sector")
    dji10y.append(df)
dji10y = pd.concat(dji10y)
dji10y["Volatility"] = (dji10y["High"] - dji10y["Low"]) / (dji10y["High"] + dji10y["Low"]) * 2
```

```python
sb.relplot(data=dji10y, x="yy-mm", y="High", errorbar=None, hue="Code", kind="line", col="Sector") 
```

```python
secs = ["Industrials", "Financial Services", "Technology"]
sb.lineplot(data=dji10y[dji10y["Sector"].isin(secs)].groupby(["yy-mm", "Sector"]).mean(numeric_only=True), 
            x="yy-mm", y="Volatility", errorbar=None, hue="Sector")
```

```python
month_mean = dji10y[dji10y["Sector"].isin(secs)].groupby(["yy-mm", "Sector"]).mean(numeric_only=True)
month_mean[month_mean["Volatility"] > 0.05]
```

```python
monthly_diff = ((dji10y.groupby(["yy-mm", "Sector"]).max("High")["High"] - dji10y.groupby("yy-mm").min("Low")["Low"])
                / (dji10y.groupby(["yy-mm", "Sector"]).max("High")["High"] + dji10y.groupby("yy-mm").min("Low")["Low"]) * 2).reset_index()
sb.lineplot(data=monthly_diff[monthly_diff["Sector"].isin(secs)], x="yy-mm", y=0, errorbar=None, hue="Sector")
```
